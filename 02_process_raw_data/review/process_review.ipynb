```python
from pathlib import Path
import pandas as pd

CURRENT_DIR = Path.cwd().resolve()
PROJECT_DIR = next((p for p in [CURRENT_DIR, *CURRENT_DIR.parents] if (p / "01_craw_data").is_dir()), CURRENT_DIR)

RAW_DIR = PROJECT_DIR / "01_craw_data" / "review"
PROCESSED_DIR = PROJECT_DIR / "02_process_raw_data" / "review"
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

print(f"Thư mục dữ liệu thô: {RAW_DIR}")
print(f"Thư mục dữ liệu sạch: {PROCESSED_DIR}")
```

## 1. Chạy tiến trình tiền xử lý đánh giá

```python
from process_review import clean_reviews

input_file = RAW_DIR / "product_review.csv"
output_file = PROCESSED_DIR / "product_review.csv"

clean_reviews(input_file, output_file, chunksize=150000)
```

## 2. Kiểm tra mẫu dữ liệu sau làm sạch

```python
df_sample = pd.read_csv(output_file, nrows=5)
display(df_sample)

print("Kiểm tra 100.000 dòng đầu tiên:")
df_100k = pd.read_csv(output_file, nrows=100000)
print("Phân bố điểm rating:")
display(df_100k["rating"].value_counts().sort_index())
print("Tỷ lệ đã mua hàng (purchased):")
display(df_100k["purchased"].value_counts(normalize=True).round(4) * 100)
```
