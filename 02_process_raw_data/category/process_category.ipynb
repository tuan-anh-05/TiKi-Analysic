```python
from pathlib import Path
import pandas as pd

# 1. Thiết lập đường dẫn
CURRENT_DIR = Path.cwd().resolve()
PROJECT_DIR = next((p for p in [CURRENT_DIR, *CURRENT_DIR.parents] if (p / "01_craw_data").is_dir()), CURRENT_DIR)

RAW_DIR = PROJECT_DIR / "01_craw_data" / "category"
PROCESSED_DIR = PROJECT_DIR / "02_process_raw_data" / "category"
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

print(f"Thư mục dữ liệu thô: {RAW_DIR}")
print(f"Thư mục dữ liệu sạch: {PROCESSED_DIR}")
```

## 1. Tiền xử lý `category.csv`

```python
from process_category import clean_categories

input_cat = RAW_DIR / "category.csv"
output_cat = PROCESSED_DIR / "category.csv"

df_cat = clean_categories(input_cat, output_cat)
display(df_cat.head(10))
print("Thông tin DataFrame:")
df_cat.info()
```

## 2. Tiền xử lý `sub_category.csv`

```python
from process_category import clean_sub_categories

input_sub = RAW_DIR / "sub_category.csv"
output_sub = PROCESSED_DIR / "sub_category.csv"

valid_ids = set(df_cat["category_id"])
df_sub = clean_sub_categories(input_sub, output_sub, valid_ids)
display(df_sub.head(10))
print("Thông tin DataFrame:")
df_sub.info()
```
